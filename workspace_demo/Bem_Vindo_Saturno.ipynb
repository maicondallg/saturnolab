```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("vendas_mensais.csv")
  df.head(8)
```

```python
# Rode a célula acima primeiro (Shift+Enter), depois aperte Ctrl+I aqui e peça:
# "plota um gráfico comparando receita e lucro por região com seaborn"

resumo = df.groupby("mes")[["receita", "lucro"]].sum().reset_index()
plt.figure(figsize=(8, 4))
sns.lineplot(data=resumo, x="mes", y="receita", marker="o", label="Receita")
sns.lineplot(data=resumo, x="mes", y="lucro", marker="s", label="Lucro")
plt.title("Evolução de Receita vs Lucro Mensal")
plt.grid(alpha=0.25)
plt.tight_layout()
plt.show()
```
